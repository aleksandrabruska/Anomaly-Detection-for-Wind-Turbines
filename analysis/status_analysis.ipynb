```python
import os
import pandas as pd
import numpy as np
```

```python
KELMARSH_PATHS = [
    os.path.join("..", "data", "Kelmarsh", "Status_Kelmarsh_1_2021-01-01_-_2021-07-01_228.csv"),
    os.path.join("..", "data", "Kelmarsh", "Status_Kelmarsh_2_2021-01-01_-_2021-07-01_229.csv"),
    os.path.join("..", "data", "Kelmarsh", "Status_Kelmarsh_3_2021-01-01_-_2021-07-01_230.csv"),
    os.path.join("..", "data", "Kelmarsh", "Status_Kelmarsh_4_2021-01-01_-_2021-07-01_231.csv"),
    os.path.join("..", "data", "Kelmarsh", "Status_Kelmarsh_5_2021-01-01_-_2021-07-01_232.csv"),
    os.path.join("..", "data", "Kelmarsh", "Status_Kelmarsh_6_2021-01-01_-_2021-07-01_233.csv")
]
PENMANSHIEL_PATHS = [
    os.path.join("..", "data", "Penmanshiel", "Status_Penmanshiel_11_2021-01-01_-_2021-07-01_1051.csv"),
    os.path.join("..", "data", "Penmanshiel", "Status_Penmanshiel_12_2021-01-01_-_2021-07-01_1052.csv"),
    os.path.join("..", "data", "Penmanshiel", "Status_Penmanshiel_13_2021-01-01_-_2021-07-01_1053.csv"),
    os.path.join("..", "data", "Penmanshiel", "Status_Penmanshiel_14_2021-01-01_-_2021-07-01_1054.csv"),
    os.path.join("..", "data", "Penmanshiel", "Status_Penmanshiel_15_2021-01-01_-_2021-07-01_1056.csv")
]
```

```python
class Overview:
    def __init__(self, path):
        self.path = path
        self.df = pd.read_csv(path, skiprows=9)
        self.name = "_".join(os.path.basename(path).split("_")[:3])
        
        self.text_to_file = ""
        self.do_steps()
        
        
    def change_dash_to_nan(self):
        self.df.replace('-', np.nan, inplace=True)
    
    def do_steps(self):
        self.change_dash_to_nan()
        
        self.save_to_text(f"""Status overview for {self.name}\n""") # Title
        
        start, end = self.calculate_start_and_end_date()
        self.save_to_text("Time range:")
        self.save_to_text(f"""\tstart date: {start}""")
        self.save_to_text(f"""\tend date: {end}""")
        
        self.save_to_text("")

        
        num_columns, num_rows = self.count_rows_and_columns()
        self.save_to_text(f"""Columns number: {num_columns}""")
        self.save_to_text(f"""Rows number: {num_rows}""")
        
        data_points = self.calculate_total_data_points()
        self.save_to_text(f"""Total data points: {data_points}""")
        
        missing_data_points = self.calculate_missing_data_points()
        self.save_to_text(f"""Missing data points: {missing_data_points}""")
        
        self.save_to_text("")
            
        missing_timestamp_start = self.calculate_missing_values_in_column('Timestamp start')
        self.save_to_text(f"""Missing timestamp start count: {missing_timestamp_start}""")
        
        missing_timestamp_end = self.calculate_missing_values_in_column('Timestamp end')
        self.save_to_text(f"""Missing timestamp end count: {missing_timestamp_end}""")
        
        missing_duration = self.calculate_missing_values_in_column('Duration')
        self.save_to_text(f"""Missing duration count: {missing_duration}""")
        
        missing_status = self.calculate_missing_values_in_column('Status')
        self.save_to_text(f"""Missing status: {missing_status}""")
        
        missing_message = self.calculate_missing_values_in_column('Message')
        self.save_to_text(f"""Missing message: {missing_message}""")
        
        missing_service_contract_category = self.calculate_missing_values_in_column('Service contract category')
        self.save_to_text(f"""Missing service_contract_category: {missing_service_contract_category}""")
        
        missing_IEC_category = self.calculate_missing_values_in_column('IEC category')
        self.save_to_text(f"""Missing IEC category: {missing_IEC_category}""")
        
        
        #comments are almost always empty
        missing_comments = self.calculate_missing_values_in_column('Comment')
        self.save_to_text(f"""Missing comments: {missing_comments}""")
        
        self.save_to_text(f"""Missing data points without comments {missing_data_points - missing_comments}""")
        
        self.save_to_file()
        
        print("done")
        
    def save_to_file(self):
        path = os.path.join("status_overviews", self.name + ".txt")
        with open(path, "w") as file:
            file.write(self.text_to_file)
        
        
    def save_to_text(self, text):
        self.text_to_file += (text + "\n")
        
        
    def count_rows_and_columns(self):
        num_columns = self.df.shape[1]
        num_rows = self.df.shape[0]
        return num_rows, num_columns
        
        
    def calculate_total_data_points(self):
        data_points = 0
        for column in self.df.columns:
            data_points += self.df[column].notna().sum()
        return data_points
        
    def calculate_missing_data_points(self):
        missing_data_points = 0
        for column in self.df.columns:
            missing_data_points += self.df[column].isna().sum()
        return missing_data_points
        
    def calculate_missing_values_in_column(self, col_name):
        return self.df[col_name].isna().sum()
    
    
    def calculate_start_and_end_date(self):
        start = self.df['Timestamp start'].iloc[0]
        end = self.df['Timestamp start'].iloc[-1]
        return start, end
    
    
    
```

```python

for i in KELMARSH_PATHS:
    Overview(i)
```

```python
for i in PENMANSHIEL_PATHS:
    Overview(i)
```
